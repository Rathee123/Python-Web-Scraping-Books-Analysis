```python
all_book_titles = []
all_book_prices = []
all_book_ratings = []
all_book_availability = []

for page in range(1, 51):
    if page == 1:
        page_url = base_url
    else:
        page_url = f"https://books.toscrape.com/catalogue/page-{page}.html"

    print(page_url)

    page_response = requests.get(page_url)
    page_soup = BeautifulSoup(page_response.text,"html.parser")

    all_book_titles.extend(
        [book.a["title"] for book in page_soup.find_all("h3")]
    )

    all_book_prices.extend(
        [price.get_text(strip=True) for price in page_soup.find_all("p", class_="price_color")]
    )

    all_book_ratings.extend(
        [rating["class"][1] for rating in page_soup.find_all("p", class_="star-rating")]
    )

    all_book_availability.extend(
        [item.get_text(strip=True) for item in page_soup.find_all("p", class_="instock")]
    )

print("Titles:",len(all_book_titles))
print("Prices:",len(all_book_prices))
print("Ratings:",len(all_book_ratings))
print("Availability:",len(all_book_availability))

```

```python
all_books_df = pd.DataFrame({
    "Book Title": all_book_titles,
    "Price": all_book_prices,
    "Rating": all_book_ratings,
    "Availability": all_book_availability
})
```
